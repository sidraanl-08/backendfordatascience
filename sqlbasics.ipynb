# 3 Major DataTypes In SQL:

1. String (14 datatypes)
- CHAR(size) -> 0 to 255, can be changed per our will
- VARCHAR(size) -> 0 to 65535, can take long paragraphs
- BINARY (size) -> same like CHAR but works with binary values
- VARBINARY(size) -> same like VARCHAR but works with binary values
- TINYTEXT() - 255, fixed characters, cant be changed
- TEXT(size) -> 65535 bytes, data being stored in bytes
- MEDIUMTEXT -> 1677215, used for long paragraphs
- LONGTEXT -> 4294967295, also used for long paragraphs
- TINYBLOB -> 255 bytes
- BLOB(size) -> 65535 bytes
- MEDIUMBLOB -> 1677215 bytes
- LONGBLOB -> 4294967295 bytes
- ENUM(val1, val2, val3) -> list of values upto 65535
- SET(val1, val2, val3) -> list of values upto 64 


2. Numeric (13 datatypes)

- BIT(size) -> 1 to 64
- TINYINT(size) -> -128 to 127
- INT(size) -> -2147483648 to 2147483647 ,abbreviation for INTEGER
- INTEGER(size) -> -2147483648 to 2147483647, same as INT
- SMALLINT(size) -> -32768 to 32767
- MEDIUMINT(size) -> -8388608 to 8388607
- BIGINT(size) -> 1212742987398131731, big value like a phone number
- BOOL -> 0/1 where 0 means false and 1 means true, abbreviation for BOOLEAN 
- BOOLEAN ->  0/1 where 0 means false and 1 means true
- FLOAT(p) -> all decimal numbers, can be fixed using (p)
- DOUBLE(size, d) -> 255.568 where size determines left side digits from dot and d determines right side digits from dot
- DECIMAL(size, d) -> size should be max 60 and d should be max 30, unlike double where it wasnt fixed
- DEC(size, d) -> same as DECIMAL, abbreviation for DECIMAL 


3. Date & Time (5 datatypes)

- DATE -> year-month-date 
- DATETIME(fsp) -> YYYY-MM-DD hh:mm:ss
- TIMESTAMP(fsp) -> gone time
- TIME (fsp) -> hh:mm:ss
- YEAR -> four digit format 2025
